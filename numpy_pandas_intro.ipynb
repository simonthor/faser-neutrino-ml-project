## When something goes wrong: debugging

A mistake in a program is called a **bug**, and finding it is called **debugging**. Every

### 1. Read the error message from the bottom up

When Python cannot run a cell, it prints an error message called a **traceback**. For
example:

```

Cell In[1], line 2
      1 momenta = [12.5, 3.1, 48.0]
----> 2 total = momenta.sum()
      3 print(total)

```

- The **last line** says what went wrong: a Python list does not have `.sum()` (a NumPy

- If the message is long, the line you wrote is usually near the bottom. The lines

## 1.1 Why NumPy?

Suppose a vertex has five primary tracks and we know the momentum of each one, in GeV.
In plain Python we would store them in a list:

````python
momenta_list = [12.5, 3.1, 48.0, 7.7, 21.4]
momenta_list * 2
````

Multiplying a list by 2 does **not** double the numbers. It sticks two copies of the list
together. To double each number in plain Python you need a loop:

````python
doubled = []
for p in momenta_list:
    doubled.append(p * 2)
doubled
````

NumPy gives us the **array**: a list of numbers that does maths the way you would expect.
It is also much faster. On a million numbers, a NumPy calculation is often more than 10
times quicker than a loop, and our data has hundreds of thousands of rows.

Everyone imports NumPy under the short name `np`:

````python
import numpy as np

momenta = np.array([12.5, 3.1, 48.0, 7.7, 21.4])
momenta * 2
````

Source: [NumPy: the absolute basics for beginners](https://numpy.org/doc/stable/user/absolute_beginners.html)

## 1.2 Creating arrays

There are several ways to make an array:

- `np.array(a_list)` turns a list into an array.
- `np.arange(start, stop, step)` counts from `start` up to, but **not including**, `stop`.
- `np.linspace(start, stop, num)` gives `num` evenly spaced numbers from `start` to
  `stop`, this time **including** `stop`.
- `np.zeros(n)` gives `n` zeros.

````python
print(np.arange(0, 10, 2))
print(np.arange(5))          # with only one number, it counts from 0
print(np.linspace(0, 1, 5))
print(np.zeros(3))
````

An array knows how many numbers it holds and what type they are:

- `len(a)` is the number of elements.
- `a.shape` is the size in each direction. For a one-dimensional array it looks like
  `(10,)`. The round brackets make it a **tuple**, which is like a list that cannot be
  changed. You will see why it is useful in section 1.7.
- `a.dtype` is the data type: `int64` for whole numbers, `float64` for decimal numbers.
  Every element of an array has the same type.

````python
plates = np.arange(1, 11)
print(plates)
print(len(plates))
print(plates.shape)
print(plates.dtype)
print(momenta.dtype)
````

Source: [Array creation](https://numpy.org/doc/stable/user/basics.creation.html)

## 1.3 Maths on whole arrays

The operators `+`, `-`, `*`, `/` and `**` (power) work **element by element**: on an
array and a number, or on two arrays of the same length. NumPy also has the usual maths
functions, such as `np.sqrt`, `np.sin`, `np.cos`, `np.arctan`, `np.log` and `np.exp`,
and they work on every element at once.

As an example, take four tracks. The **transverse momentum** $p_T$ is the part of the
momentum perpendicular to the beam. If $p_x$ and $p_y$ are the two components
perpendicular to the beam, then $p_T = \sqrt{p_x^2 + p_y^2}$:

````python
px = np.array([0.3, -1.2, 0.8, 0.05])   # GeV
py = np.array([0.4, 0.5, -0.6, 0.1])    # GeV

pt = np.sqrt(px**2 + py**2)
pt
````

One line, no loop, and it would work just as well for a million tracks.

Source: [Mathematical functions](https://numpy.org/doc/stable/reference/routines.math.html)

## 1.4 Slicing: taking a piece

You already know that `a_list[2]` gives the element at position 2, counting from 0.
**Slicing** gives a whole piece at once: `a_list[start:stop]` gives the elements from
position `start` up to, but not including, `stop`.

- Leave out `start` to begin at the start: `a_list[:3]`.
- Leave out `stop` to go on to the end: `a_list[3:]`.
- Negative numbers count from the end: `a_list[-1]` is the last element.
- A third number is the step: `a_list[::2]` takes every second element.

A good way to remember it is to imagine the positions as fence posts *between* the
elements. `letters[1:4]` is everything between post 1 and post 4:

```
  | a | b | c | d | e | f |
  0   1   2   3   4   5   6
```

````python
letters = ["a", "b", "c", "d", "e", "f"]

print(letters[1:4])
print(letters[:3])
print(letters[3:])
print(letters[-1])
print(letters[-2:])
print(letters[::2])
````

NumPy arrays are sliced in exactly the same way:

````python
energies = np.array([5.0, 12.0, 30.0, 2.5, 80.0, 41.0, 9.0])

print(energies[2:5])
print(energies[-3:])
````

Source: [Lists, in the Python tutorial](https://docs.python.org/3/tutorial/introduction.html#lists)

## 1.5 Asking yes/no questions: boolean masks

Comparing an array with a number gives an array of `True` and `False`, one per element.
`True` and `False` are called **booleans**. Putting such an array inside the square brackets
keeps only the elements where it is `True`. This is called a **mask**, because it covers
up the elements you do not want, like a stencil.

````python
is_high = energies > 20
print(is_high)
print(energies[is_high])
````

Masks can be combined:

| symbol | meaning | example |
|---|---|---|
| `&` | and | `(energies > 5) & (energies < 50)` |
| `\|` | or | `(energies < 5) \| (energies > 50)` |
| `~` | not | `~is_high` |

The round brackets around each comparison are **required**. Without them Python does the
`&` first and you get a confusing error. The words `and`, `or` and `not` do not work on
arrays either; use the symbols.

````python
print(energies[(energies > 5) & (energies < 50)])
print(energies[(energies < 5) | (energies > 50)])
print(energies[~is_high])
````

One more trick: in maths, `True` counts as 1 and `False` as 0. So the sum of a mask is
*how many* elements are `True`, and the mean is *what fraction* of them are `True`.

````python
print(is_high.sum())
print(is_high.mean())
````

Source: [Boolean array indexing](https://numpy.org/doc/stable/user/basics.indexing.html#boolean-array-indexing)

## 1.6 Summary numbers

Often we want a single number that describes a whole array:

- `a.sum()`, `a.mean()`, `a.min()`, `a.max()`
- `np.median(a)`: the **median** is the middle value once the numbers are sorted, so half
  of the values are below it and half above. One huge value changes the mean a lot, but
  hardly moves the median. (It is written `np.median(a)`, because `a.median()` does not
  exist.)
- `a.std()`: the **standard deviation** says how spread out the values are. Roughly, it
  is the typical distance of a value from the mean.
- `a.argmax()` and `a.argmin()`: the *position* of the largest or smallest value.

````python
print(momenta)
print(f"mean   = {momenta.mean():.1f} GeV")
print(f"median = {np.median(momenta):.1f} GeV")
print(f"std    = {momenta.std():.1f} GeV")
print(f"the largest momentum is at position {momenta.argmax()}")
````

Source: [Statistics](https://numpy.org/doc/stable/reference/routines.statistics.html)

## 1.7 Tables of numbers: 2D arrays

An array can have rows and columns, like a table. You make one from a list of lists, where
each inner list is one row. Here are four tracks, each with its momentum components
$p_x$, $p_y$ and $p_z$ (along the beam):

````python
tracks = np.array([
    [0.3, 0.4, 25.0],
    [-1.2, 0.5, 60.0],
    [0.8, -0.6, 12.0],
    [0.05, 0.1, 3.0],
])   # one row per track; the columns are px, py, pz in GeV

print(tracks.shape)   # (rows, columns)
````

To pick out elements you now give two positions, the row and the column:
`tracks[row, column]`. Each of them can be a single number, a slice, or a list.

| code | gives |
|---|---|
| `tracks[1, 2]` | one number: row 1, column 2 |
| `tracks[1]` | all of row 1 (the same as `tracks[1, :]`) |
| `tracks[:, 2]` | all of column 2. A `:` on its own means "everything" |
| `tracks[1:3, 0:2]` | rows 1 and 2, columns 0 and 1 |
| `tracks[[0, 2]]` | rows 0 and 2. A **list** of positions picks out exactly those, in that order |
| `tracks[:, [0, 2]]` | columns 0 and 2 |

Notice the double square brackets in `tracks[[0, 2]]`. The outer pair means "pick from
`tracks`", the inner pair is the list `[0, 2]`.

````python
print(tracks[1, 2])
print(tracks[:, 2])
print(tracks[1:3, 0:2])
print(tracks[[0, 2]])
````

Masks work on rows too. `tracks[:, 2] > 10` is `True` for each track whose $p_z$ is
above 10 GeV, and using it as the row selection keeps those tracks:

````python
fast = tracks[:, 2] > 10
tracks[fast]
````

### Summary numbers on 2D arrays: `axis`

`tracks.sum()` adds up *every* number in the table, which is rarely what you want. The
`axis` option chooses the direction:

- `axis=0` works down the rows, giving one result per **column**.
- `axis=1` works across the columns, giving one result per **row**.

A way to remember it: the axis you name is the one that disappears. `tracks` has shape
`(4, 3)`. `tracks.sum(axis=0)` has shape `(3,)` and `tracks.sum(axis=1)` has shape `(4,)`.

````python
total = tracks.sum(axis=0)   # add the tracks together: the total momentum of the vertex
print(total)

largest_component = tracks.max(axis=1)   # one number per track
print(largest_component)
````

## 1.8 Missing values: NaN

Sometimes a number could not be measured. In FASERnu, for example, a track that is too
short has no measured momentum. NumPy marks a missing number as `np.nan`, short for
"not a number".

Any maths involving NaN gives NaN, so the mean of an array with a single NaN in it is
also NaN. The functions `np.nanmean`, `np.nansum`, `np.nanmax`, `np.nanmedian` and so on
skip the NaNs. `np.isnan` gives a mask that is `True` where the value is missing.

````python
measured = np.array([12.5, np.nan, 48.0, 7.7, np.nan])

print(measured.mean())
print(np.nanmean(measured))
print(np.isnan(measured))
print(measured[~np.isnan(measured)])
````

Be careful: you **cannot** find NaNs with `==`. NaN is not equal to anything, not even to
itself, so `measured == np.nan` is always `False`. Always use `np.isnan`.

````python
measured == np.nan
````

### Changing values with a mask

A mask can also be used on the left of `=`, to change only the selected elements. Here
every momentum above 20 GeV is replaced by 20:

````python
capped = momenta.copy()
capped[capped > 20] = 20

print(momenta)
print(capped)
````

Why `.copy()`? Writing `capped = momenta` would not make a new array. It would only give
the *same* array a second name, so changing `capped` would change `momenta` as well.
`.copy()` makes a separate array.

Source: [numpy.nanmean](https://numpy.org/doc/stable/reference/generated/numpy.nanmean.html)

## 2.1 Series and DataFrame

pandas has two main objects:

- a **Series** is one column: a list of values, each with a label;
- a **DataFrame** is a table: several Series side by side, sharing the same labels.

A quick way to make a DataFrame is from a **dictionary**. A dictionary is written with
curly brackets and stores pairs `key: value`. Here each key is a column name and each value
is the list of entries in that column. This small table is made up, but it has the same
columns as the real data:

## 2.5 `.loc`: choosing rows and columns together

`.loc[rows, columns]` picks rows and columns at the same time. The rows can be a mask, an
index label or `:` (everything). The columns can be a name or a list of names.

````python
small.loc[small["n_vtrk"] > 6, "p_mean"]
````

````python
small.loc[small["n_vtrk"] > 6, ["interaction", "p_mean"]]
````

Note the difference from `.iloc`: `.loc` uses the **labels** of the index, and `.iloc` uses
the **positions**. For `small` they are the same, but after filtering or sorting they are
not (you will see an example in section 2.9).

`.loc` is also how you **change** values. Here every `p_mean` above 30 GeV is replaced by
30 in a copy of the table:

````python
small_capped = small.copy()
small_capped.loc[small_capped["p_mean"] > 30, "p_mean"] = 30
small_capped
````

As with NumPy arrays (section 1.8), `.copy()` is needed. `small_capped = small` would only
give the same table a second name.

> **⚠️ Do not trust old code from Stack Overflow**
>
> Many answers on [Stack Overflow](https://stackoverflow.com/), and many older tutorials,
> change values like this:
>
> ```python
> small[small["p_mean"] > 30]["p_mean"] = 30     # does NOT work
> ```
>
> This is called **chained assignment**: two pairs of square brackets one after the other.
> The first part, `small[...]`, makes a new, separate table, and the second part changes
> *that* table, which is then thrown away. Since pandas 3.0, which this project uses,
> this **never** changes the original. pandas prints a `ChainedAssignmentError` warning,
> but the data stays as it was. Always change values with one
> `.loc[rows, column] = value`.
>
> Code from Stack Overflow often looks right, but it may have been written for a much
> older version of pandas and be wrong today. If code you found behaves strangely, check
> the official documentation. `pd.__version__` tells you which version you have.

Sources: [Indexing and selecting data](https://pandas.pydata.org/docs/user_guide/indexing.html),
[Copy-on-Write](https://pandas.pydata.org/docs/user_guide/copy_on_write.html)

## 2.8 `groupby`: one answer per group

`groupby` splits the rows into groups by the value of a column, does a calculation on
each group, and puts the answers together in one table. "The mean number of tracks for
each interaction" is a single line:

````python
small.groupby("interaction")["n_vtrk"].mean()
````

`.agg` does several calculations at once:

````python
small.groupby("interaction")["n_vtrk"].agg(["mean", "median", "count"])
````

### A word about `weight`

Each simulated vertex stands for a number of real vertices that FASER expects to see
(see the README). That number is the `weight` column. So the **sum** of `weight` over a
group is the number of real vertices of that kind we expect. `weight` must never be used
as an input to a machine learning model, because it gives the answer away; the README
explains why.

## 2.10 Missing values

- `isna()` gives a mask that is `True` where a value is missing.
- `dropna()` removes every row with a missing value. `dropna(subset=["p_mean"])` only looks
  at the listed columns.
- `fillna(value)` fills in the missing values.

Unlike NumPy's `.mean()`, pandas' `.mean()`, `.sum()` and so on **skip** missing values
automatically.

````python
print(small["p_mean"].isna())
print(small["p_mean"].mean())
````

````python
small.dropna(subset=["p_mean"])
````

````python
small["p_mean"].fillna(0)
````

### Filling with another column

`fillna` can also be given a whole Series instead of a single number. Then each missing
value is filled with the value from the **same row** of the other Series. pandas matches
the rows by their index labels.

Here is an example from the real data. `delta_phi_p_mean` weights each track by its
momentum, so it is missing when fewer than two tracks have a measured momentum.
`delta_phi_mean` counts every track equally, so it always exists. One possible fill is
to use the unweighted value wherever the weighted one is missing:

````python
delta_phi_p_filled = df["delta_phi_p_mean"].fillna(df["delta_phi_mean"])

print(f"missing before: {df['delta_phi_p_mean'].isna().sum()}")
print(f"missing after : {delta_phi_p_filled.isna().sum()}")
````

````python
missing_rows = df["delta_phi_p_mean"].isna()

comparison = df.loc[missing_rows, ["delta_phi_mean", "delta_phi_p_mean"]]
comparison["filled"] = delta_phi_p_filled[missing_rows]
comparison.head()
````

Whether filling in is a good idea at all is a physics question, not a programming one.
There is always a *reason* why a value is missing (`explore.ipynb` has more on this).

Source: [Working with missing data](https://pandas.pydata.org/docs/user_guide/missing_data.html)
